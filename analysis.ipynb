```python
#import
from datetime import datetime as dt
import pandas as pd
# 1. Gender Sales Analysis
df = pd.read_csv('Retail Sales Dataset.csv')
gender_more_buy = df.groupby('Gender')['Total Amount'].sum()
pergentage_gender = (gender_more_buy.iloc[0]) / (gender_more_buy.iloc[0] + gender_more_buy.iloc[1]) * 100
print(gender_more_buy.iloc[0]),
print(gender_more_buy.iloc[1]),
print( pergentage_gender)
```

Female sales are slightly higher than male sales, but the distribution is very close, showing a balanced customer base between genders (Females 51% vs. Males 49%).

```python
## 2. Monthly Sales Trend Analysis
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')
df['month_benefits'] = df['Date'].dt.month
df['year_benefits'] = df['Date'].dt.year
month_year_sales= df.groupby(['year_benefits', 'month_benefits'])['Total Amount'].sum()
print(month_year_sales)
```

Peak Month: May 2023 recorded the highest sales reaching $56,450.

Lowest Month (2023): September recorded the lowest sales in 2023 at $23,620.

Note on 2024 Data: Sales for January 2024 appear very low ($1,530) due to incomplete data recording for that year in the dataset

```python
## 3. Age Analysis per Category
mean_age_product_gategory = df.groupby('Product Category')['Age'].mean()
print(mean_age_product_gategory)
```

The average age across all product categories is nearly identical (around 40 to 41 years old). This indicates that the products aren't targeting a specific age group, and the store's customer base is fairly balanced.

```python
more_quant = df.groupby('Product Category')['Quantity'].sum()
print(more_quant)
```

Clothing is the top-performing category in terms of quantity sold with 939 units, followed by Electronics with 882 units, while Beauty came in last with 811 units.

```python
more_city = df.groupby('City')['Total Price'].sum()
print(more_city)
```

San Francisco is by far the top-performing city in overall revenue, crossing $71,000. On the other hand, New York is lagging behind as the lowest-selling city (around $34,000), which might need better marketing efforts or operational reviews.

## Executive Summary
Key takeaways from the analysis:
1. Sales are almost evenly split between genders (Females 51% vs. Males 49%).
2. May 2023 was the peak sales month, while September 2023 was the lowest.
3. Clothing is the best-selling category by volume (units sold).
4. San Francisco is the strongest market, while New York needs performance improvement.
